# Hybrid recommender

Quick demo: build the content and collaborative engines, combine them in the hybrid
engine, and get recommendations. The `w_content` weight sets the balance between them.

```python
import pandas as pd
import sys
sys.path.append('..')

from src.recommenders.content_based import ContentBasedRecommender
from src.recommenders.collaborative_filtering import CollaborativeFilteringRecommender
from src.recommenders.hybrid_recommender import HybridRecommender

tracks = pd.read_csv('../data/raw/tracks.csv')
playlists = pd.read_csv('../data/raw/playlists.csv')

content_rec = ContentBasedRecommender(tracks)
collab_rec = CollaborativeFilteringRecommender(playlists)
hybrid_rec = HybridRecommender(content_rec, collab_rec)
print('Hybrid engine ready')
```

```python
# Mix content and collaborative scores; w_content leans toward audio similarity
sample_id = tracks['track_id'].iloc[0]
recs = hybrid_rec.get_recommendations(sample_id, w_content=0.6, top_n=5)
print(recs)
```
