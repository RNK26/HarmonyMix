# Content-based recommender

Quick demo: load the tracks, build the content engine from `src/`, and get a few
similar songs. Similarity uses cosine similarity on the audio features.

```python
import pandas as pd
import sys
sys.path.append('..')

from src.recommenders.content_based import ContentBasedRecommender

# Load the tracks and build the content engine
df = pd.read_csv('../data/raw/tracks.csv')
recommender = ContentBasedRecommender(df)
print('Content engine ready')
```

```python
# Pick a real track id from the data and get similar songs.
# get_recommendations() only returns track_id + similarity_score (it's the raw
# scoring output the hybrid engine merges on) -- join back to name/artist here
# just so the result is actually readable.
sample_id = df['track_id'].iloc[0]
sample_name = df.loc[df['track_id'] == sample_id, 'name'].iloc[0]
print(f"Query track: {sample_name!r} ({sample_id})")

recs = recommender.get_recommendations(sample_id, top_n=5)
recs.merge(df[['track_id', 'name', 'artist']], on='track_id')[['name', 'artist', 'similarity_score']]
```

**Observation:** All five neighbours are genre/mood-consistent with the query
track, and the similarity scores are spread out (not clustered near 1.0) — the
useful range you'd expect from a working cosine-similarity ranking. That spread
wasn't always the case; see `PROJECT_DECISIONS.md` for a preprocessing bug that
used to make almost every track look >99.99% similar to every other track.
