# Collaborative recommender

Quick demo: build the collaborative engine from the listening data and get songs
that co-occur with a chosen track in people's playlists.

```python
import pandas as pd
import sys
sys.path.append('..')

from src.recommenders.collaborative_filtering import CollaborativeFilteringRecommender

playlists = pd.read_csv('../data/raw/playlists.csv')
tracks = pd.read_csv('../data/raw/tracks.csv', usecols=['track_id', 'name', 'artist'])
recommender = CollaborativeFilteringRecommender(playlists)
print('Collaborative engine ready')
```

```python
# Use a track that exists in the listening data.
# get_recommendations() only returns track_id + collaborative_score, so join
# back to name/artist (loaded above) to make the result human-readable.
sample_id = recommender.tracks[0]
sample_name = tracks.loc[tracks['track_id'] == sample_id, 'name'].iloc[0]
print(f"Query track: {sample_name!r} ({sample_id})")

recs = recommender.get_recommendations(sample_id, top_n=5)
recs.merge(tracks, on='track_id')[['name', 'artist', 'collaborative_score']]
```

**Observation:** `collaborative_score` here is a normalized co-occurrence count
(Section 9 of `EDA.ipynb`), not a similarity in the audio sense — two tracks can
score highly just because they're frequently played back-to-back, regardless of
how different they sound. That's the intentional difference from content-based
filtering, and exactly why the hybrid engine keeps both signals instead of
picking one.
